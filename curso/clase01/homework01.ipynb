## Ejercicio 1 · Clasificar una edad (decisión)

Implementa `clasificar(edad)` que devuelva:
- `"menor"` si `edad < 18`
- `"adulto"` si `18 <= edad < 65`
- `"adulto mayor"` si `edad >= 65`

**Casos de ejemplo:** `clasificar(10)` → `"menor"`, `clasificar(40)` → `"adulto"`, `clasificar(70)` → `"adulto mayor"`.

```python
def clasificar(edad):
    if edad < 18:
      return "menor"

    elif edad < 65:
      return "adulto"

    else:
      return "adulto mayor"

```

```python
# === Tests visibles · Ejercicio 1 ===
print (clasificar(10)) == 'menor'
print (clasificar(40))== 'adulto'
print (clasificar(70))== 'adulto mayor'
print("✅ Ejercicio 1: tests visibles superados.")
```

## Ejercicio 2 · Suma de los pares hasta n (bucle + condición)

Implementa `suma_pares(n)` que devuelva la suma de todos los números **pares** desde 1 hasta `n` (inclusive).

**Casos de ejemplo:** `suma_pares(6)` → `12` (2+4+6); `suma_pares(1)` → `0`.

```python
def suma_pares(n):
    suma = 0

    for numero in range (1,n+1):

       if numero % 2 == 0:

        suma = suma + numero

    return suma
```

```python
# === Tests visibles · Ejercicio 2 ===
print (suma_pares(6)) == 12
print (suma_pares(1)) == 0
print("✅ Ejercicio 2: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 2 ===
print (suma_pares(2)) == 2
print (suma_pares(10)) == 30
print (suma_pares(0)) == 0
print("✅ Ejercicio 2: tests adicionales superados.")
```

## Ejercicio 3 · Factorial (acumulador multiplicativo)

Implementa `factorial(n)` que devuelva `n!` = 1·2·3·…·n. Por convención, `factorial(0)` → `1`.

**Casos de ejemplo:** `factorial(5)` → `120`; `factorial(0)` → `1`.

```python
def factorial(n):

    resultado = 1

    for numero in range(1,n+1):

        resultado = resultado *numero

    return resultado
```

```python
# === Tests visibles · Ejercicio 3 ===
print(factorial(5)) == 120
print (factorial(0)) == 1
print("✅ Ejercicio 3: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 3 ===
print(factorial(1)) == 1
print(factorial(3)) == 6
print (factorial(7))== 5040
print("✅ Ejercicio 3: tests adicionales superados.")
```

## Ejercicio 4 · Contar vocales (recorrer una cadena)

Implementa `contar_vocales(texto)` que cuente las vocales (`a, e, i, o, u`) en un texto **en minúsculas**.

**Casos de ejemplo:** `contar_vocales("murcielago")` → `5`; `contar_vocales("xyz")` → `0`.

```python
def contar_vocales(texto):

    contador = 0

    for letras in texto:
      if letras in "aeiou":
        contador = contador + 1

    return contador
```

## Ejercicio 6 · Invertir una lista (sin reverse ni [::-1])

Implementa `invertir(lista)` que devuelva una **nueva** lista con los elementos en orden inverso, **sin** usar `lista[::-1]` ni `.reverse()`.

**Casos de ejemplo:** `invertir([1, 2, 3])` → `[3, 2, 1]`; `invertir([])` → `[]`.

```python
def invertir(lista):
   nueva = []

   for i in range(len(lista)-1,-1,-1):
    nueva.append(lista[i])

   return nueva
```

## Ejercicio 7 · ¿Es primo? (bucle + salida temprana)

Implementa `es_primo(n)` que devuelva `True` si `n` es primo y `False` si no. Recuerda: los primos son enteros `>= 2` divisibles solo por 1 y por sí mismos. `0` y `1` no son primos.

**Casos de ejemplo:** `es_primo(7)` → `True`; `es_primo(1)` → `False`; `es_primo(9)` → `False`.

```python
def es_primo(n):

    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True
```

```python
# === Tests visibles · Ejercicio 7 ===
print (es_primo(7)) is True
print (es_primo(1)) is False
print (es_primo(9)) is False
print("✅ Ejercicio 7: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 7 ===
print (es_primo(2)) is True
print (es_primo(0)) is False
print (es_primo(97)) is True
print (es_primo(100)) is False
print("✅ Ejercicio 7: tests adicionales superados.")
```

## Ejercicio 8 · La moda: el valor más frecuente (contador)

Implementa `moda(lista)` que devuelva el elemento que **más veces** aparece en una lista no vacía. Si hay empate, devuelve el que aparece primero al recorrer la lista.

**Casos de ejemplo:** `moda([1, 2, 2, 3])` → `2`; `moda([5])` → `5`.

```python
def moda(lista):

    moda_actual = lista [0]
    max_repeticiones = lista.count(moda_actual)

    for numero in lista:
        repeticiones = lista.count(numero)

        if repeticiones > max_repeticiones:
            moda_actual = numero
            max_repeticiones = repeticiones

    return moda_actual
```

```python
# === Tests visibles · Ejercicio 8 ===
print (moda([1, 2, 2, 3])) == 2
print (moda([5])) == 5
print("✅ Ejercicio 8: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 8 ===
print (moda([4, 4, 4, 1, 1])) == 4
print (moda(['a', 'b', 'a'])) == 'a'
print (moda([7, 8, 8, 7, 7])) == 7
print("✅ Ejercicio 8: tests adicionales superados.")
```
