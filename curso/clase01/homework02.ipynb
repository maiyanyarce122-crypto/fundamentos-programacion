## Ejercicio 1 · Parsear las notas (de texto a datos)

Las notas llegan como un texto: `"Ana:4.5,Luis:3.0,Sara:2.5"`.

Implementa `parsear(texto)` que devuelva una **lista de tuplas** `(nombre, nota)` donde el nombre es texto y la nota es `float`.

**Ejemplo:** `parsear("Ana:4.5,Luis:3.0")` → `[('Ana', 4.5), ('Luis', 3.0)]`.

Pista: `texto.split(",")` separa por comas; `parte.split(":")` separa nombre y nota; `float(...)` convierte a número.

````python
def parsear(texto):
    resultado =[]

    estudiantes = texto.split(",")

    for estudiante in estudiantes:
        nombre, nota = estudiante.split(":")
        resultado.append((nombre, float(nota)))

    return resultado
````

````python
# === Tests visibles · Ejercicio 1 ===
print (parsear('Ana:4.5,Luis:3.0')) == [('Ana', 4.5), ('Luis', 3.0)]
print(("✅ Ejercicio 1: tests visibles superados."))
````

````python
# === Tests adicionales (ocultos) · Ejercicio 1 ===
print(parsear('Sara:2.5') == [('Sara', 2.5)])
print (parsear('A:5.0,B:0.0,C:3.3')) == [('A', 5.0), ('B', 0.0), ('C', 3.3)]
print("✅ Ejercicio 1: tests adicionales superados.")
````

## Ejercicio 2 · Promedio del curso (acumulador)

Implementa `promedio(notas)` que reciba una lista de tuplas `(nombre, nota)` y devuelva el promedio de las notas como `float`. Si la lista está vacía, devuelve `None`.

**Ejemplo:** `promedio([('Ana', 4.0), ('Luis', 2.0)])` → `3.0`.

````python
def promedio(notas):

    if len(notas) == 0:
      return None

    suma = 0

    for nombre, nota in notas:
      suma = suma + nota

    promedio = suma / len(notas)

    return promedio
````

````python
# === Tests visibles · Ejercicio 2 ===
print (promedio([('Ana', 4.0), ('Luis', 2.0)])) == 3.0
print (promedio([])) is None
print("✅ Ejercicio 2: tests visibles superados.")
````

````python
# === Tests adicionales (ocultos) · Ejercicio 2 ===
print (promedio([('X', 5.0)])) == 5.0
print((promedio([('A', 3.0), ('B', 4.0), ('C', 5.0)]) - 4.0) < 1e-9)
print("✅ Ejercicio 2: tests adicionales superados.")
````

## Ejercicio 3 · El mejor estudiante (patrón campeón)

Implementa `mejor(notas)` que devuelva el **nombre** del estudiante con la nota más alta. En caso de empate, el primero que aparece.

**Ejemplo:** `mejor([('Ana', 4.5), ('Luis', 3.0)])` → `'Ana'`.

````python
def mejor(notas):
    mejor_nombre = notas[0][0]
    mejor_nota = notas[0][1]

    for nombre, nota in notas:
        if nota > mejor_nota:
            mejor_nota = nota
            mejor_nombre = nombre

    return mejor_nombre
````

````python
# === Tests visibles · Ejercicio 3 ===
print(mejor([('Ana', 4.5), ('Luis', 3.0)])) == 'Ana'
print("✅ Ejercicio 3: tests visibles superados.")
````

````python
# === Tests adicionales (ocultos) · Ejercicio 3 ===
print (mejor([('A', 2.0), ('B', 4.0), ('C', 4.0)])) == 'B'
print (mejor([('Solo', 1.0)])) == 'Solo'
print("✅ Ejercicio 3: tests adicionales superados.")
````

## Ejercicio 4 · Cuántos aprobaron (contador con umbral)

Implementa `aprobados(notas, umbral=3.0)` que cuente cuántos estudiantes tienen nota **mayor o igual** al umbral.

**Ejemplo:** `aprobados([('A', 3.0), ('B', 2.9), ('C', 4.0)])` → `2`.

````python
def aprobados(notas, umbral=3.0):
    contador = 0

    for nombre, nota in notas:
        if nota >= umbral:
            contador = contador + 1

    return contador
````

````python
# === Tests visibles · Ejercicio 4 ===
print (aprobados([('A', 3.0), ('B', 2.9), ('C', 4.0)])) == 2
print("✅ Ejercicio 4: tests visibles superados.")
````

````python
# === Tests adicionales (ocultos) · Ejercicio 4 ===
print (aprobados([('A', 1.0), ('B', 2.0)])) == 0
print (aprobados([('A', 4.0), ('B', 5.0)], umbral=4.5)) == 1
print("✅ Ejercicio 4: tests adicionales superados.")
````

## Ejercicio 5 · Distribución por bandas (varios contadores + decisión)

Implementa `distribucion(notas)` que devuelva una tupla `(reprobados, regulares, buenos)` contando:
- **reprobados:** nota < 3.0
- **regulares:** 3.0 <= nota < 4.0
- **buenos:** nota >= 4.0

**Ejemplo:** `distribucion([('A', 2.0), ('B', 3.5), ('C', 4.8)])` → `(1, 1, 1)`.

````python
def distribucion(notas):
    reprobados = 0
    regulares = 0
    buenos = 0

    for nombre, nota in notas:
        if nota < 3.0:
            reprobados = reprobados + 1
        elif nota < 4.0:
            regulares = regulares + 1
        else:
            buenos = buenos + 1

    return (reprobados, regulares, buenos)
````

````python
# === Tests visibles · Ejercicio 5 ===
print  (distribucion([('A', 2.0), ('B', 3.5), ('C', 4.8)])) == (1, 1, 1)
print("✅ Ejercicio 5: tests visibles superados.")
````

````python
# === Tests adicionales (ocultos) · Ejercicio 5 ===
print (distribucion([])) == (0, 0, 0)
print (distribucion([('A', 2.9), ('B', 3.0), ('C', 3.9), ('D', 4.0)])) == (1, 2, 1)
print("✅ Ejercicio 5: tests adicionales superados.")
````

## Ejercicio 6 · Integración: resumen del curso

¡La pieza final! Implementa `resumen(texto)` que reciba el texto crudo de notas y devuelva una tupla con cuatro valores, **reutilizando tus funciones anteriores**:

`(cantidad_estudiantes, promedio_curso, mejor_estudiante, num_aprobados)`

**Ejemplo:** `resumen("Ana:4.0,Luis:2.0,Sara:5.0")` → `(3, 3.6666..., 'Sara', 2)`.

Esto demuestra el poder de **descomponer**: ya tienes todas las piezas; solo las combinas.

````python
def resumen(texto):
    notas = parsear(texto)

    cantidad_estudiantes = len(notas)
    promedio_curso = promedio(notas)
    mejor_estudiante = mejor(notas)
    num_aprobados = aprobados(notas)

    return (cantidad_estudiantes, promedio_curso, mejor_estudiante, num_aprobados)
````

````python
# === Tests visibles · Ejercicio 6 ===
_r = resumen('Ana:4.0,Luis:2.0,Sara:5.0')
print (_r[0]) == 3
print (abs(_r[1] - 11.0/3) < 1e-9)
print (_r[2]) == 'Sara'
print (_r[3]) == 2
print("✅ Ejercicio 6: tests visibles superados.")
````

````python
# === Tests adicionales (ocultos) · Ejercicio 6 ===
_r2 = resumen('A:1.0,B:1.5')
print (_r2 == (2, 1.25, 'B', 0))
print("✅ Ejercicio 6: tests adicionales superados.")
````
