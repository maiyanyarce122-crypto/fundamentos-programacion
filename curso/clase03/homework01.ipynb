## Ejercicio 1 · Máximo Común Divisor (algoritmo de Euclides)

Implementa `mcd(a, b)` que devuelva el Máximo Común Divisor de dos enteros positivos usando el **algoritmo de Euclides**: mientras `b != 0`, reemplaza `(a, b)` por `(b, a % b)`; cuando `b == 0`, devuelve `a`.

**Ejemplos:** `mcd(48, 18)` → `6`; `mcd(100, 75)` → `25`; `mcd(13, 7)` → `1`.

Inicio

Mientras b sea diferente de 0

    Guardar el valor de a % b

    Reemplazar a por b

    Reemplazar b por el residuo

Fin del ciclo

Devolver a

Fin

```python
def mcd(a, b):
    while b != 0:
      a,b = b, a % b

    return a
```

```python
# === Tests visibles · Ejercicio 1 ===
print( mcd(48, 18)) == 6
print( mcd(100, 75)) == 25
print( mcd(13, 7)) == 1
print("✅ Ejercicio 1: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 1 ===
print( mcd(1, 1)) == 1
print( mcd(252, 105)) == 21
print( mcd(17, 17)) == 17
print( mcd(36, 24)) == 12
print("✅ Ejercicio 1: tests adicionales superados.")
```

## Ejercicio 2 · Bubble sort (ordenamiento propio)

Implementa `bubble_sort(lista)` que ordene una lista de números de **menor a mayor** usando el algoritmo de burbuja: recorre la lista repetidamente e intercambia elementos adyacentes que estén en el orden incorrecto. Devuelve una **nueva lista** (no modifica la original).

**Ejemplo:** `bubble_sort([5, 2, 4, 1, 3])` → `[1, 2, 3, 4, 5]`.

Pista: una pasada completa 'burbujea' el mayor al final. Repite `n-1` pasadas.

Inicio

Crear una copia de la lista

Recorrer la lista n - 1 veces

    Recorrer la lista hasta la parte que aún no está ordenada

        Si el elemento actual es mayor que el siguiente

            Intercambiar ambos elementos

Devolver la copia de la lista ordenada

Fin

```python
def bubble_sort(lista):
    nueva_lista = lista.copy()

    n = len(nueva_lista)

    for i in range(n - 1):
        for j in range(n - i - 1):
            if nueva_lista[j] > nueva_lista[j + 1]:
                nueva_lista[j], nueva_lista[j + 1] = (nueva_lista[j + 1], nueva_lista[j])

    return nueva_lista
```

```python
# === Tests visibles · Ejercicio 2 ===
print( bubble_sort([5, 2, 4, 1, 3])) == [1, 2, 3, 4, 5]
print( bubble_sort([1])) == [1]
print( bubble_sort([])) == []
print("✅ Ejercicio 2: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 2 ===
print( bubble_sort([3, 3, 1])) == [1, 3, 3]
print( bubble_sort([-2, 0, -5, 3])) == [-5, -2, 0, 3]
_orig = [4, 2, 1]; bubble_sort(_orig); print( _orig) == [4, 2, 1]
print("✅ Ejercicio 2: tests adicionales superados.")
```

## Ejercicio 3 · Contar ocurrencias de cada letra

Implementa `contar_letras(texto)` que reciba una cadena (solo letras minúsculas y espacios) y devuelva una **lista de tuplas** `(letra, conteo)` ordenada **alfabéticamente** por letra. Ignora los espacios.

**Ejemplo:** `contar_letras("abba")` → `[('a', 2), ('b', 2)]`.

Pista: construye dos listas paralelas (letras únicas + conteos), luego ordena.

Inicio

Crear una lista para guardar las letras únicas

Crear una lista para guardar el conteo de cada letra

Recorrer cada carácter del texto

    Si el carácter es un espacio
        Continuar con el siguiente

    Si la letra no está en la lista de letras
        Agregar la letra a la lista
        Agregar 1 a la lista de conteos

    Si la letra ya existe
        Buscar su posición
        Aumentar el conteo en 1

Ordenar las letras alfabéticamente junto con sus conteos

Crear una lista de tuplas

Devolver la lista de tuplas

Fin

```python
def contar_letras(texto):
    letras = []
    conteos = []

    for letra in texto:
       if letra == "":
        continue

       if letra not in letras:
        letras.append(letra)
        conteos.append(1)
       else:
           indice = letras.index(letra)
           conteos[indice] += 1


    resultado = []

    for i in range(len(letras)):
        resultado.append((letras[i], conteos[i]))

    resultado.sort()

    return resultado
```

```python
# === Tests visibles · Ejercicio 3 ===
print( contar_letras('abba')) == [('a', 2), ('b', 2)]
print( contar_letras('aaa')) == [('a', 3)]
print("✅ Ejercicio 3: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 3 ===
print( contar_letras('')) == []
print( contar_letras('cba')) == [('a', 1), ('b', 1), ('c', 1)]
print( contar_letras('hola mundo')) == [('a', 1), ('d', 1), ('h', 1), ('l', 1), ('m', 1), ('n', 1), ('o', 2), ('u', 1)]
print("✅ Ejercicio 3: tests adicionales superados.")
```

## Ejercicio 4 · Números primos hasta n (criba manual)

Implementa `primos_hasta(n)` que devuelva una **lista** con todos los números primos desde 2 hasta `n` inclusive. Un número es primo si solo es divisible por 1 y por sí mismo.

**Ejemplo:** `primos_hasta(20)` → `[2, 3, 5, 7, 11, 13, 17, 19]`.

Pista: escribe una función auxiliar `es_primo(k)` y usa un bucle externo.

Inicio

Si k es menor que 2
    Devolver False

Recorrer los números desde 2 hasta k-1

    Si k es divisible por alguno de ellos
        Devolver False

Devolver True

Fin

Inicio

Crear una lista vacía

Recorrer los números desde 2 hasta n

    Si el número es primo
        Agregarlo a la lista

Devolver la lista

Fin

```python
def es_primo(k):
    if k < 2:
        return False

    for i in range(2, k):
        if k % i == 0:
            return False

    return True

def primos_hasta(n):
    lista_primos = []

    for numero in range(2, n + 1):
        if es_primo(numero):
          lista_primos.append(numero)

    return lista_primos
```

```python
# === Tests visibles · Ejercicio 4 ===
print( primos_hasta(20)) == [2, 3, 5, 7, 11, 13, 17, 19]
print( primos_hasta(2)) == [2]
print( primos_hasta(1)) == []
print("✅ Ejercicio 4: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 4 ===
print( primos_hasta(10)) == [2, 3, 5, 7]
print( 97 in primos_hasta(100))
print( 100 not in primos_hasta(100))
print("✅ Ejercicio 4: tests adicionales superados.")
```

## Ejercicio 5 · Aplicar descuentos con map y lambda

Implementa `aplicar_descuentos(precios, tasa)` que reciba una lista de precios y una tasa de descuento (ej: `0.10` = 10%) y devuelva una **nueva lista** con los precios tras el descuento, usando `map` y `lambda`.

Los precios resultado deben ser enteros (usa `int()` en la lambda).

**Ejemplo:** `aplicar_descuentos([100000, 200000], 0.10)` → `[90000, 180000]`.

Inicio

Recibir la lista de precios

Recibir la tasa de descuento

Aplicar el descuento a cada precio usando map y lambda

Convertir cada resultado a entero

Devolver la nueva lista

Fin

```python
def aplicar_descuentos(precios, tasa):
    return list(map(lambda precio: int(precio*(1-tasa)),precios))
```

```python
# === Tests visibles · Ejercicio 5 ===
print( aplicar_descuentos([100000, 200000], 0.10)) == [90000, 180000]
print( aplicar_descuentos([50000], 0.50)) == [25000]
print("✅ Ejercicio 5: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 5 ===
print( aplicar_descuentos([], 0.20)) == []
print( aplicar_descuentos([10000, 20000, 30000], 0.0)) == [10000, 20000, 30000]
print( aplicar_descuentos([99999], 0.19)) == [80999]
print("✅ Ejercicio 5: tests adicionales superados.")
```

## Ejercicio 6 · Potencias de 2 hasta 2^n

Implementa `potencias_de_2(n)` que devuelva una **lista** con las potencias de 2 desde 2^0 hasta 2^n inclusive.

**Ejemplo:** `potencias_de_2(5)` → `[1, 2, 4, 8, 16, 32]`.

La lista tiene `n + 1` elementos.

Inicio

Crear una lista vacía

Recorrer los números desde 0 hasta n

    Calcular 2 elevado al número actual

    Agregar el resultado a la lista

Devolver la lista

Fin

```python
def potencias_de_2(n):
    potencias = []

    for i in range(n+1):
        potencias.append(2** i)

    return potencias
```

```python
# === Tests visibles · Ejercicio 6 ===
print( potencias_de_2(5)) == [1, 2, 4, 8, 16, 32]
print( potencias_de_2(0)) == [1]
print("✅ Ejercicio 6: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 6 ===
print( potencias_de_2(1)) == [1, 2]
print( potencias_de_2(10)[-1]) == 1024
print( len(potencias_de_2(8))) == 9
print("✅ Ejercicio 6: tests adicionales superados.")
```

## Ejercicio 7 · Acumular comisión por tramos de ventas

Implementa `comision_total(ventas)` que reciba una lista de montos de venta y devuelva la **comisión total acumulada**. La tasa de comisión depende de cada venta individual:
- venta < 500.000 → 5%
- 500.000 <= venta <= 2.000.000 → 8%
- venta > 2.000.000 → 12%

**Ejemplo:** `comision_total([200000, 800000, 3000000])` → `10000 + 64000 + 360000 = 434000`.

nicio

Crear una variable comision en 0

Recorrer cada venta de la lista

    Si la venta es menor que 500000
        Sumar el 5% de la venta

    Si la venta está entre 500000 y 2000000
        Sumar el 8% de la venta

    Si la venta es mayor que 2000000
        Sumar el 12% de la venta

Devolver la comisión total

Fin

```python
def comision_total(ventas):
    comision = 0

    for venta in ventas:
        if venta < 500000:
            comision += venta * 0.05
        elif venta <= 2000000:
            comision += venta * 0.08
        else:
            comision += venta * 0.12

    return comision
```

```python
# === Tests visibles · Ejercicio 7 ===
print( comision_total([200000, 800000, 3000000])) == 434000.0
print(comision_total([])) == 0
print("✅ Ejercicio 7: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 7 ===
print( comision_total([100000])) == 5000.0
print( comision_total([500000])) == 40000.0
print( comision_total([2000001])) == 240000.12
print( abs(comision_total([499999, 500000, 2000000, 2000001]) - (499999*0.05 + 500000*0.08 + 2000000*0.08 + 2000001*0.12)) < 0.01)
print("✅ Ejercicio 7: tests adicionales superados.")
```

## Ejercicio 8 · Validar condición sobre todos/alguno de los elementos

Implementa dos funciones:
- `todos_cumplen(lista, condicion)`: devuelve `True` si **todos** los elementos de la lista cumplen `condicion`. Lista vacía → `True`.
- `alguno_cumple(lista, condicion)`: devuelve `True` si **al menos un** elemento cumple `condicion`. Lista vacía → `False`.

Implementa internamente con un **bucle** (no uses `all()` ni `any()` directamente en la función principal).

**Ejemplos:**
`todos_cumplen([2, 4, 6], lambda x: x % 2 == 0)` → `True`
`alguno_cumple([1, 3, 4], lambda x: x % 2 == 0)` → `True`

inicio - todos cumplen

Recorrer cada elemento de la lista

    Si algún elemento NO cumple la condición

        Devolver False

Si termina el recorrido

    Devolver True

Fin

Inicio alguno cumple

Recorrer cada elemento de la lista

    Si algún elemento cumple la condición

        Devolver True

Si termina el recorrido

    Devolver False

Fin

```python
def todos_cumplen(lista, condicion):
    for elemento in lista:
      if not condicion(elemento):
        return False

    return True

def alguno_cumple(lista, condicion):
    for elemento in lista:
        if condicion(elemento):
           return True

    return False
```

```python
# === Tests visibles · Ejercicio 8 ===
print( todos_cumplen([2, 4, 6], lambda x: x % 2 == 0)) is True
print( todos_cumplen([2, 3, 6], lambda x: x % 2 == 0)) is False
print( alguno_cumple([1, 3, 4], lambda x: x % 2 == 0)) is True
print( alguno_cumple([1, 3, 5], lambda x: x % 2 == 0)) is False
print("✅ Ejercicio 8: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 8 ===
print( todos_cumplen([], lambda x: False)) is True
print( alguno_cumple([], lambda x: True)) is False
print( todos_cumplen([10, 20, 30], lambda x: x > 5)) is True
print( alguno_cumple([-1, -2, 3], lambda x: x > 0)) is True
print("✅ Ejercicio 8: tests adicionales superados.")
```
