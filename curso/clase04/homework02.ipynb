## Ejercicio 1 · Cargar el catálogo

El catálogo de productos ya está definido en la celda de identificación como la variable `CATALOGO` (lista de dicts). Implementa `ids_por_categoria(catalogo)` que devuelva un dict `{categoria: set_de_ids}`.

**Ejemplo:**
```
ids_por_categoria(CATALOGO)['tecnologia']
# -> {'P001', 'P002', 'P003', 'P008'}
```

Esta es la estructura de **índice por categoría** que permite recuperar todos los productos de una categoría en O(1).

````python
CATALOGO = [ {'id': 'P001', 'nombre': 'laptop', 'categoria': 'tecnologia', 'precio': 2_500_000, 'stock': 15}, {'id': 'P002', 'nombre': 'teclado', 'categoria': 'tecnologia', 'precio': 120_000, 'stock': 45}, {'id': 'P003', 'nombre': 'monitor', 'categoria': 'tecnologia', 'precio': 800_000, 'stock': 8}, {'id': 'P004', 'nombre': 'silla', 'categoria': 'hogar', 'precio': 450_000, 'stock': 20}, {'id': 'P005', 'nombre': 'escritorio','categoria': 'hogar', 'precio': 900_000, 'stock': 5}, {'id': 'P006', 'nombre': 'arroz', 'categoria': 'alimentos', 'precio': 5_000, 'stock': 200}, {'id': 'P007', 'nombre': 'aceite', 'categoria': 'alimentos', 'precio': 18_000, 'stock': 150}, {'id': 'P008', 'nombre': 'mouse', 'categoria': 'tecnologia', 'precio': 80_000, 'stock': 60}, {'id': 'P009', 'nombre': 'lampara', 'categoria': 'hogar', 'precio': 75_000, 'stock': 30}, {'id': 'P010', 'nombre': 'cafe', 'categoria': 'alimentos', 'precio': 25_000, 'stock': 80}, ]
````

Estructura de datos: Diccionario

````python
def ids_por_categoria(catalogo):
    indice = {}

    for producto in catalogo:
        categoria = producto["categoria"]
        id_producto = producto["id"]

        if categoria not in indice:
            indice[categoria] = set()

        indice[categoria].add(id_producto)

    return indice
````

## Ejercicio 2 · Índice id → producto

Implementa `indice_por_id(catalogo)` que devuelva un dict `{id: producto}` donde `producto` es el dict completo del producto.

Este índice permite recuperar todos los datos de un producto por su id en O(1),
en vez de recorrer el catálogo completo cada vez.

**Ejemplo:**
`indice_por_id(CATALOGO)['P003']['nombre']` → `'monitor'`

````python
def indice_por_id(catalogo):
    indice = {}

    for producto in catalogo:
        indice[producto["id"]] = producto

    return indice
````

````python
# === Tests visibles · Ejercicio 2 ===
print (indice_por_id(CATALOGO)['P003']['nombre']) == 'monitor'
print (indice_por_id(CATALOGO)['P006']['precio']) == 5_000
print("✅ Ejercicio 2: tests visibles superados.")
````

````python
# === Tests adicionales (ocultos) · Ejercicio 2 ===
print (set(indice_por_id(CATALOGO).keys())) == {p['id'] for p in CATALOGO}
print (indice_por_id(CATALOGO)['P001']['stock']) == 15
print (indice_por_id([])) == {}
print("✅ Ejercicio 2: tests adicionales superados.")
````

## Ejercicio 3 · Buscar por categoría

Implementa `buscar_categoria(catalogo, categoria)` que devuelva una **lista de dicts** con todos los productos de esa categoría.

Usa las funciones `ids_por_categoria` e `indice_por_id` que ya implementaste.

El resultado debe estar **ordenado por precio ascendente**.

Si la categoría no existe, devuelve lista vacía.

````python
def buscar_categoria(catalogo, categoria):
    categorias = ids_por_categoria(catalogo)

    if categoria not in categorias:
        return []

    productos = []
    indice = indice_por_id(catalogo)

    for id_producto in categorias[categoria]:
      productos.append(indice[id_producto])

    productos.sort(key=lambda producto: producto["precio"])

    return productos
````

````python
# === Tests visibles · Ejercicio 3 ===
_alim = buscar_categoria(CATALOGO, 'alimentos')
print( len(_alim)) == 3
print(  _alim[0]['nombre']) == 'arroz'
print(  _alim[-1]['nombre']) == 'cafe'
print("✅ Ejercicio 3: tests visibles superados.")
````

````python
# === Tests adicionales (ocultos) · Ejercicio 3 ===
_tec = buscar_categoria(CATALOGO, 'tecnologia')
print(  [p['id'] for p in _tec]) == ['P008', 'P002', 'P003', 'P001']
print(  buscar_categoria(CATALOGO, 'deportes')) == []
print("✅ Ejercicio 3: tests adicionales superados.")
````

## Ejercicio 4 · Buscar por rango de precio

Implementa `buscar_rango_precio(catalogo, precio_min, precio_max)` que devuelva una **lista de dicts** con los productos cuyo precio esté entre `precio_min` y `precio_max` (inclusive).

El resultado debe estar ordenado por precio ascendente.

````python
def buscar_rango_precio(catalogo, precio_min, precio_max):
    productos = []

    for producto in catalogo:
        if precio_min <= producto["precio"] <= precio_max:
            productos.append(producto)

    productos.sort(key=lambda producto:producto["precio"])

    return productos
````

````python
# === Tests visibles · Ejercicio 4 ===
_rango = buscar_rango_precio(CATALOGO, 50_000, 200_000)
print( len(_rango)) == 3
print( _rango[0]['nombre']) == 'lampara'
print("✅ Ejercicio 4: tests visibles superados.")
````

## Ejercicio 5 · Búsqueda multi-filtro

Implementa `buscar(catalogo, categoria=None, precio_min=None, precio_max=None, en_stock=False)` que aplique **todos** los filtros que no sean `None`:

- `categoria`: si se especifica, solo productos de esa categoría.
- `precio_min`: si se especifica, precio >= precio_min.
- `precio_max`: si se especifica, precio <= precio_max.
- `en_stock=True`: solo productos con stock > 0.

Devuelve lista de dicts ordenada por precio ascendente.

Usa las funciones anteriores como bloques de construcción o filtra directamente.

````python
def buscar(catalogo, categoria=None, precio_min=None, precio_max=None, en_stock=False):
    if categoria is not None:
       productos = buscar_categoria(catalogo, categoria)

    else:
      productos = catalogo.copy()

    if precio_min is not None:
      productos = [producto for producto in productos if producto["precio"] >= precio_min]

    if precio_max is not None:
      productos = [producto for producto in productos if producto["precio"] <= precio_max]

    if en_stock:
      productos = [producto for producto in productos if producto["stock"] > 0]

    productos.sort(key=lambda producto: producto["precio"])

    return productos
````

````python
# === Tests visibles · Ejercicio 5 ===
_r1 = buscar(CATALOGO, categoria='hogar')
print( len(_r1)) == 3
_r2 = buscar(CATALOGO, precio_min=100_000, precio_max=500_000)
print( all(100_000 <= p['precio'] <= 500_000 for p in _r2))
print("✅ Ejercicio 5: tests visibles superados.")
````

## Ejercicio 6 · Resumen del catálogo

Implementa `resumen_catalogo(catalogo)` que devuelva un dict con el
resumen estadístico del catálogo, **reutilizando las funciones anteriores**:

```python
{

  'categorias': set,                # conjunto de categorías únicas
  'precio_min': int,                # precio del producto más barato
  'precio_max': int,                # precio del producto más caro
  'producto_mas_barato': str,       # nombre del producto de menor precio
  'productos_en_stock': int,        # cantidad con stock > 0
}
```

Si el catálogo está vacío, devuelve el dict con 0, set vacío, None donde corresponda.

````python
def resumen_catalogo(catalogo):
    if len(catalogo) ==0:
      return {'total_productos': 0, 'categorias': set(), 'precio_min': None, 'precio_max': None, 'producto_mas_barato': None, 'productos_en_stock': 0}

    productos_ordenados = buscar_rango_precio(catalogo, min(producto["precio"] for producto in catalogo), max(producto["precio"] for producto in catalogo))

    categorias = set(producto["categoria"] for producto in catalogo)



    return {
        'total_productos': len(catalogo),
        'categorias': categorias,
        'precio_min': productos_ordenados[0]["precio"],
        'precio_max': productos_ordenados[-1]["precio"],
        'producto_mas_barato': productos_ordenados[0]["nombre"],
        'productos_en_stock': len(buscar(catalogo, en_stock=True))}
````
